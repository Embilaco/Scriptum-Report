```python
REPORT_DIR = Path(os.getcwd())
#print(os.getcwd(),REPORT_DIR)

TEST_ROOT = REPORT_DIR.parent.parent
if str(TEST_ROOT) not in sys.path:
    sys.path.append(str(TEST_ROOT))

from tests.baseTestRoot import *

import Scriptum # type: ignore
mydoc = None

```

```python

def base_setup(tmp_path):

    setupTestEnvironment(tmp_path, Path("data"), REPORT_DIR, 
                         include_patterns = [
                            "*.rdf",
                            "template.docx",
                         ])
    os.chdir(tmp_path)
    print(tmp_path)
```

```python
%%ipytest -vvv -s
def test_me(tmp_path):
    base_setup(tmp_path)

    rdf = Scriptum.ReportDataFile('word_input.rdf')
    if rdf.errors: 
        # assemble collected errors if any
        print('\n'.join(rdf.errors))
    #rdf.inspect()
    #rdf.showFiles()

    mydoc = Scriptum.ManagedDocx('template.docx')
    print(mydoc.errors, mydoc.warnings)

    mydoc.typesetting(rdf,addcopy=True, directfill=True, globalfill=True, cleanup=True, removetemplate=True, cleardust=True, setproperties=True)

    mydoc.save('result_word.docx', finish=True, createpdf=True)

    props =  ['author','category', 'comments', 'created', 
            'identifier', 'keywords', 'language', 'last_modified_by', 'modified',
            'last_printed', 'revision', 'subject', 'title', 'version' ] 
    # skipped some
    for prop in props:
        print("prop %r, value %r"%(prop,getattr(mydoc.document.core_properties,prop)))

    print(tmp_path)
```
